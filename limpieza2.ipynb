```python
import pandas as pd
from nycflights13 import flights, airports, airlines 
```

```python
fl_air = pd.merge(flights, airports, left_on="dest", right_on="faa", how = "left")
```

```python
fl_air.info()
```

```python
fl_air.describe(include="all")
```

```python
fl_air["sched_dep_time"].isna()
```

```python
fl_air = fl_air.dropna(subset=["dep_time", "arr_time"])
```

```python
# encuentra el delay máximo
numero_que_llena = fl_air["dep_delay"].max() 
# con el delay máximo reemplazo lo vacío con él
fl_air["dep_delay"].fillna(numero_que_llena)
```
