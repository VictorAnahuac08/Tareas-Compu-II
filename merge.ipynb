# Merge (Joins) en pandas

```python
import pandas as pd
from nycflights13 import flights , airlines , planes , airports , weather
```

```python
# sólo me inetersa los meses enero, febrero y marzo
df1 = flights[(flights["month"] == 1) | (flights["month"] == 2) | (flights["month"] == 3) ]
# sólo el aeropuerto jfk
df1 = df1[df1["origin"]=="JFK"] 
# aquellos vuelos con delay
df1 = df1[df1["dep_delay"]>0]
df1.head()
# limpiar vacíos 
df1.dropna()
```

```python
print(df1.merge( 
    planes , 
    on = "tailnum" , how = "inner").shape[0]
)
```
